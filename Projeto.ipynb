# Brain Hemorrhage Extended (BHX): Bounding box extrapolation from thick to thin slice CT images

It was created to support:

- Brain hemorrhage detection
- Medical image annotation,
- Biomedical data analysis,
- Artificial intelligence research using CT scans.

The dataset contains:

CT imaging metadata,
hemorrhage annotations,
bounding box information,
manually labeled cases,
and extrapolated annotations.

All CT images with 3 slices or more.

CT Scan Images

        ↓

Radiologist Annotation

        ↓

Bounding Boxes

        ↓

Manual Label CSV

        ↓

Automatic Extrapolation

        ↓

Expanded Annotation Dataset
